```python
import pandas as pd
import numpy as np

# === BAGIAN 1: MEMUAT & MEMBERSIHKAN DATA ===
df = pd.read_csv('Amazon Sale Report.csv', low_memory=False)

cols_to_drop = ['index', 'Unnamed: 22']
df_clean = df.drop(columns=[c for c in cols_to_drop if c in df.columns])
df_clean = df_clean.drop_duplicates(subset=['Order ID', 'SKU'])
df_clean = df_clean.dropna(subset=['Amount', 'Category', 'ship-state'])
df_clean = df_clean[df_clean['Amount'] > 0]
df_clean['Date'] = pd.to_datetime(df_clean['Date'], format='%m-%d-%y', errors='coerce')
df_clean = df_clean.dropna(subset=['Date'])
df_clean['ship-state'] = df_clean['ship-state'].str.upper().str.strip()

# === BAGIAN 2: STATISTIK DESKRIPTIF & OUTLIER ===
total_penjualan = df_clean['Amount'].sum()
rata_transaksi = df_clean['Amount'].mean()
median_transaksi = df_clean['Amount'].median()
total_unit_terjual = df_clean['Qty'].sum()
kategori_terlaris = df_clean.groupby('Category')['Amount'].sum().idxmax()
wilayah_teratas = df_clean.groupby('ship-state')['Amount'].sum().idxmax()
status_terbanyak = df_clean['Status'].value_counts().idxmax()

print("=======================================================")
print("   LAPORAN ANALISIS DATA PENJUALAN E-COMMERCE AMAZON   ")
print("=======================================================")
print(f"1. Total Penjualan       : INR {total_penjualan:,.2f}")
print(f"2. Rata-rata Transaksi   : INR {rata_transaksi:,.2f}")
print(f"3. Median Transaksi      : INR {median_transaksi:,.2f}")
print(f"4. Total Produk Terjual  : {total_unit_terjual:,} unit")
print(f"5. Kategori Terlaris     : {kategori_terlaris}")
print(f"6. Wilayah Teratas       : {wilayah_teratas}")
print(f"7. Status Pesanan Utama  : {status_terbanyak}")
print("=======================================================\n")

# Deteksi Outlier (IQR)
Q1 = df_clean['Amount'].quantile(0.25)
Q3 = df_clean['Amount'].quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outliers = df_clean[(df_clean['Amount'] < batas_bawah) | (df_clean['Amount'] > batas_atas)]

print("=== DETEKSI OUTLIER (METODE IQR) ===")
print(f"Q1 (25%)          : INR {Q1:,.2f}")
print(f"Q3 (75%)          : INR {Q3:,.2f}")
print(f"IQR               : INR {IQR:,.2f}")
print(f"Batas Atas Normal : INR {batas_atas:,.2f}")
print(f"Jumlah Outlier    : {len(outliers)} transaksi ({len(outliers)/len(df_clean)*100:.2f}% dari total data)")
print(f"Transaksi Tertinggi (Outlier Ekstrem): INR {df_clean['Amount'].max():,.2f}")
```

```python
# GRAFIK 1: BAR CHART (Total Penjualan per Kategori)
plt.figure(figsize=(10, 5))
cat_sales = df_clean.groupby('Category')['Amount'].sum().sort_values(ascending=False)
sns.barplot(x=cat_sales.index, y=cat_sales.values, palette='viridis')
plt.title('Total Penjualan Berdasarkan Kategori Produk', fontsize=14, fontweight='bold')
plt.xlabel('Kategori Produk')
plt.ylabel('Total Penjualan (INR)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# GRAFIK 2: LINE CHART (Tren Penjualan Harian)
plt.figure(figsize=(12, 5))
daily_sales = df_clean.groupby('Date')['Amount'].sum()
plt.plot(daily_sales.index, daily_sales.values, color='#1f77b4', linewidth=2)
plt.title('Tren Penjualan Harian Selama Periode Pengamatan', fontsize=14, fontweight='bold')
plt.xlabel('Tanggal')
plt.ylabel('Total Penjualan Harian (INR)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# GRAFIK 3: SCATTER PLOT (Hubungan Kuantitas Pembelian vs Nilai Transaksi)
plt.figure(figsize=(10, 5))
# Mengambil sampel 3000 data agar titik scatter terlihat jelas dan tidak menumpuk
sample_df = df_clean.sample(n=min(3000, len(df_clean)), random_state=42)
sns.scatterplot(data=sample_df, x='Qty', y='Amount', hue='Fulfilment', alpha=0.6)
plt.title('Hubungan Jumlah Barang (Qty) terhadap Total Harga (Amount)', fontsize=14, fontweight='bold')
plt.xlabel('Kuantitas Barang per Transaksi (Unit)')
plt.ylabel('Nilai Transaksi (INR)')
plt.tight_layout()
plt.show()

# GRAFIK 4: HISTOGRAM (Distribusi Nilai Transaksi)
plt.figure(figsize=(10, 5))
sns.histplot(df_clean['Amount'], bins=40, kde=True, color='teal')
plt.axvline(rata_transaksi, color='red', linestyle='--', label=f'Rata-rata: {rata_transaksi:.0f}')
plt.title('Distribusi Frekuensi Nilai Transaksi Pelanggan', fontsize=14, fontweight='bold')
plt.xlabel('Nilai Transaksi (INR)')
plt.ylabel('Frekuensi Transaksi')
plt.legend()
plt.tight_layout()
plt.show()

# GRAFIK 5: BOXPLOT (Deteksi Outlier Nilai Transaksi per Kategori)
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_clean, x='Category', y='Amount', palette='Set2')
plt.title('Sebaran & Deteksi Outlier Nilai Transaksi per Kategori', fontsize=14, fontweight='bold')
plt.xlabel('Kategori Produk')
plt.ylabel('Nilai Transaksi (INR)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
